```python
#Save constants as variables

h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12
```

```python
#Put them in a data structure

constants = [h, c, pi, hbar, e, me, ep0]
print(constants)
```

```python
#Calculate the fine structure constant

alpha = (e**2)/(4*pi*ep0*hbar*c)
print(alpha)
```

```python
#Create matrix

A = [[3.14, 42], [2.71, 11]]

#Define the two column and find the determinant
column_1 = [sublist[0] for sublist in A]
column_2 = [sublist[1] for sublist in A]

det_A = column_1[0]*column_2[1] - column_2[0]*column_1[1]
print(det_A)
```

```python
#Find and print the inverse matrix

inv_A = [[column_2[1]/det_A, - column_2[0]/det_A], [-column_1[1]/det_A, column_1[0]/det_A]]
print(inv_A)
```
