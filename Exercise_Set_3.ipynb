```python
def greater_three(list_of_numbers):
    my_list = []
    for i in list_of_numbers:
        if len(i) > 3:
            my_list.append(i)
    return(my_list)

print(greater_three(["dog", "bunny", "bird", "cat", "horse", "pet"]))
```

```python
def my_fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return my_fibonacci(n-1) - my_fibonacci(n-2)

print(my_fibonacci(5))
```

```python
def the_maximum(number_list):
    the_max = number_list[0]
    for i in number_list:
        if i>the_max:
            the_max=i
    return the_max

print(the_maximum([189, 172837, 12, 378, 9834]))
```

```python
def alphabetize(word_list):
    for i in range(len(word_list)):
        for j in range(i+1, len(word_list)):
            if word_list[i] > word_list[j]:
                word_list[i], word_list[j] = word_list[j], word_list[i]
    return word_list

print(alphabetize(["hello", "goodbye", "dog", "avocado"]))
```

```python
class BankAccount:
    def __init__(self, holder_name, balance):
        self.holder_name = holder_name
        self.balance = balance
        
    def deposit(self, amount_deposited):
        self.balance += amount_deposited
        
    def withdraw(self, amount_withdrawn):
        self.balance -= amount_withdrawn
        
    def check_balance(self):
        return self.balance

my_account = BankAccount("Lola", 500)
my_account.withdraw(150)
print(my_account.check_balance())

my_account.deposit(80)
print(my_account.check_balance())
```

```python
class Library:
    def __init__(self, book_name):
        self.book_name = book_name
        self.books = []

    def add_book(self, book_added):
        self.books.append(book_added)

    def remove_book(self, book_removed):
        if book_removed in self.books:
            self.books.remove(book_removed)

    def display_books(self):
        print(f"Books in {self.book_name}:")
        for book in self.books:
            print(f" - {book}")

lolas_library = Library("Lola's Library")
lolas_library.add_book("The Starless Sea")
lolas_library.add_book("Nerve Damage")
lolas_library.add_book("The Final Empire")
lolas_library.add_book("Cleopatra and Frankestein")
lolas_library.display_books()

lolas_library.remove_book("Nerve Damage")
lolas_library.display_books()
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    def display_info(self):
        print(f"Vehicle information: Year - {self.year}, Make - {self.make}, Model - {self.model}")

class Car(Vehicle):
    def __init__(self, make, model, year, number_doors):
        super().__init__(make, model, year)
        self.number_doors = number_doors

    def display_info(self):
        super().display_info()
        print(f"Number of doors: {self.number_doors}")

class Truck(Vehicle):
    def __init__(self, make, model, year, payload_capacity):
        super().__init__(make, model, year)
        self.payload_capacity = payload_capacity

    def display_info(self):
        super().display_info()
        print(f"Payload capacity: {self.payload_capacity}")

my_vehicle = Vehicle("Honda", "Civic", 2020)
my_vehicle.display_info()

my_car = Car("Toyota", "Camry", 2021, 4)
my_car.display_info()

my_truck = Truck("Ford", "F-150", 2019, 2)
my_truck.display_info()
```

```python
class Particle:
    def __init__(self, mass, charge, position):
        self.mass = mass
        self.charge = charge
        self.position = position

    def kin_E(self, velocity):
        return 0.5*self.mass*(velocity**2)

    def pot_E(self, height):
        g=9.81
        return self.mass*g*height

my_particle = Particle(1, 1, (0, 0, 0))
print(my_particle.kin_E(5))
print(my_particle.pot_E(5))
```

```python
class ElectricField:
    def __init__(self, strength, direction):
        self.strength = strength
        self.direction = direction

    def force_on_particle(self, particle):
        fx = self.strength * self.direction[0] * particle.charge
        fy = self.strength * self.direction[1] * particle.charge
        fz = self.strength * self.direction[2] * particle.charge
        return (fx, fy, fz)

my_field = ElectricField(10, (1, 0, 0))
my_particle = Particle(1, 1, (0, 0, 0))

print(my_field.force_on_particle(my_particle))
```

```python
class Wave:
    def __init__(self, amplitude, frequency, wavelength):
        self.amplitude = amplitude
        self.frequency = frequency
        self.wavelength = wavelength

    def wave_speed(self):
        return self.frequency * self.wavelength

    def wave_energy(self):
        return 0.5*(self.amplitude**2)*(self.frequency**2)

my_wave = Wave(1, 1, 1)
print(my_wave.wave_speed())
print(my_wave.wave_energy())
```

```python
class QuantumState:
    def __init__(self, wavefunction, energy_level):
        self.wavefunction = wavefunction
        self.energy_level = energy_level

    def probability_density(self, position):
        psi = self.wavefunction(position)
        return abs(psi)**2

    def expectation_value_position(self, position_range):
        dx = position_range[1] - position_range[0]
        integral = sum(self.probability_density(x) * x * dx for x in position_range)
        return integral

import numpy as np
wavefunction = lambda x: (2/np.pi)**0.25 * np.exp(-x**2)
quantum_state = QuantumState(wavefunction, 1)

print(quantum_state.probability_density(0))
print(quantum_state.expectation_value_position(np.linspace(-5, 5, 1000)))
```

```python
class Circuit:
    def __init__(self, resistance, capacitance, inductance):
        self.resistance = resistance
        self.capacitance = capacitance
        self.inductance = inductance

    def impendace(self, frequency):
        omega = 2 * np.pi * frequency
        z_capacitive = 1 / (1j*omega*self.capacitance)
        z_inductive = 1j*omega*self.inductance
        return self.resistance + z_capacitive + z_inductive

    def resonant_frequency(self):
        return 1/(2*np.pi*(self.inductance*self.capacitance)**0.5)

my_circuit = Circuit(100, 1e-6, 1e-3)
print(my_circuit.impendace(1000))
print(my_circuit.resonant_frequency())
```

```python
class QuantumParticle:
    def __init__(self, mass, charge, spin):
        self.mass = mass
        self.charge = charge
        self.spin = spin

    def de_broglie(self, momentum):
        h = 6.62607015e-34
        return h/momentum

    def energy(self, frequency):
        h = 6.62607015e-34
        return h*frequency

class Electron(QuantumParticle):
    def __init__(self):
        super().__init__(mass=9.10938356e-31, charge=-1.602176634e-19, spin=0.5)

class Proton(QuantumParticle):
    def __init__(self):
        super().__init__(mass=1.6726219e-27, charge=1.602176634e-19, spin=0.5)
        self.atomic_number=1

electron = Electron()
proton = Proton()

print("Electron de Broglie wavelength:", electron.de_broglie(1e-24))
print("Proton energy at 1e14 Hz:", proton.energy(1e14))
```

```python
class QuantumSystem:
    def __init__(self, hamiltonian, wavefunction):
        self.hamiltonian = hamiltonian
        self.wavefunction = wavefunction

    def time_evolution(self, time):
        pass

    def expectation_value(self, observable, position_range):
        dx = position_range[1] - position_range[0]
        integral = sum(self.wavefunction(x).conjugate() * observable(x) * self.wavefunction(x) * dx for x in position_range)
        return integral

class HarmonicOscillator(QuantumSystem):
    def __init__(self, mass, frequency):
        hamiltonian = lambda x: 0.5*mass*(frequency**2)*(x**2)
        wavefunction = lambda x: (mass*frequency/np.pi)**0.25 * (2.71828)**(-mass*frequency*(x**2)/2)
        super().__init__(hamiltonian, wavefunction) 

class ParticleInBox(QuantumSystem):
    def __init__(self, box_length):
        hamiltonian = lambda x: 0 if 0 < x < box_length else float('inf')
        wavefunction = lambda x: (2 / box_length)**0.5 * (3.14159 / box_length)**0.5 * (3.14159 * x / box_length)
        super().__init__(hamiltonian, wavefunction)

ho = HarmonicOscillator(mass=9.11e-31, frequency=1e14)
pib = ParticleInBox(box_length=1e-9)

print("Harmonic Oscillator expectation value:", ho.expectation_value(lambda x: x, np.linspace(-1e-9, 1e-9, 1000)))
print("Particle in Box expectation value:", pib.expectation_value(lambda x: x, np.linspace(0, 1e-9, 1000)))
```
