```python
decimal_input = input("Enter a decimal number: ")
decimal = int(decimal_input)
```

```python
binary = bin(decimal)
print("Original binary string:", binary)

#Remove 0b

Binary = binary[2:]
print("Binary string without 0b:", Binary)

#Modify the length
if len(Binary) < 10:
    formatted_binary = Binary.zfill(10)
elif len(Binary) == 10:
    formatted_binary = Binary
elif (Binary) > 10:
    formatted_binary = Binary[-10:]

print("Binary string without 0b and filled with extra 0s to have 10 digits:", formatted_binary)
```

```python
#Repeat this with an arbitrary amount other than 10

if len(Binary) < 27:
    new_formatted_binary = Binary.zfill(27)
elif len(Binary) == 27:
    new_formatted_binary = Binary
elif (Binary) > 27:
    new_formatted_binary = Binary[-27:]

print("Binary string without 0b and filled with extra 0s to have 27 digits:", new_formatted_binary)
```

```python
#We can calculate the largest decimal number that can be represented in binary with 27 digits
# using the formula (2^n)-1

largest_number = (2**27)-1

print("The largest number that can be represented in binary with 27 digits is", largest_number)

bina = bin(largest_number)
print("In binary this number would be", bina[2:])
```

```python
#First generate the input

N = input("Enter a positive integer:")
n = int(N)
```

```python
#Now calculate the sum

result = 0

for i in range (1, n+1):
    result = result + i

print(result)
```

```python
formula = n*(n+1)/2

if result == formula:
    i = True
else:
    i = False

print("The values obtaining adding regularly, and using the formula are the same:", i)
```

```python
user_input = input("Enter a positive integer:")
my_number = int(user_input)
```

```python
reversed_number = 0

while my_number > 0:
    last_digit = my_number % 10
    reversed_number = (reversed_number * 10) + last_digit
    my_number = my_number // 10

print("Reversed number:", reversed_number)
```

```python
primes_list = [i for i in range(2, 100) if all(i%j!=0 for j in range(2, int(i**0.5)+1))]
print(primes_list)
```
