```python
#Defining the variables

h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12

#Calculating the Bohr radius

a_0 = (4*pi*ep0*hbar**2)/(me*e**2)
print("The Bohr radius is a_0=", a_0, "m")
```

```python
#Make a variable for E_1

E_1 = -(e**2)/(8*pi*ep0*a_0)
print(E_1)
```

```python
#Change E_1 to electron volts (eV) by deviding by e

E_1_eV = E_1/e
print(E_1_eV)
```

```python
#Make a list with only E_1

energy_values = [E_1_eV]
print(energy_values)
```

```python
#Define a function for E_n (in eV)

def E_(n):
    return E_1_eV/n**2

#Append E_2 to E_6 to our list
E_2, E_3, E_4, E_5, E_6 = E_(2), E_(3), E_(4), E_(5), E_(6)

energy_values.append(E_2)
energy_values.append(E_3)
energy_values.append(E_4)
energy_values.append(E_5)
energy_values.append(E_6)

print(energy_values)
```

```python
#Turn list into set

energy_values_set = set(energy_values)
print(energy_values_set)
```

```python
#Turn set back into list

e_values = list(energy_values_set)
print(e_values)
```

```python
#Sort list

e_values.sort()
print(e_values)
```

```python
#Calculate wavelengths

E_balmer_first = E_3 - E_2
wave_balmer_first = (h*c)/E_balmer_first
print(wave_balmer_first)

E_balmer_second = E_4 - E_2
wave_balmer_second = (h*c)/E_balmer_second
print(wave_balmer_second)
```

```python
#Convert back to SI by dividing the wavelengths by e

SI_wave_balmer_first = wave_balmer_first/e
print(SI_wave_balmer_first)

SI_wave_balmer_second = wave_balmer_second/e
print(SI_wave_balmer_second)
```
