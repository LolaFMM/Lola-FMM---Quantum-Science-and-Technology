```python
#Convert a decimal integer to a binary string

def decimal_to_binary(n):
    if n == 0:
        return "0"
    binary_str = ""
    while n > 0:
        bit = n % 2
        binary_str = str(bit) + binary_str
        n = n // 2
    return binary_str

decimal = 745
binary = decimal_to_binary(decimal)
print(binary)
```

```python
#Convert binary into decimal

binary = '10111001'
decimal = 0

for i in binary:
    decimal = decimal * 2 + int(i)

print(decimal)
```

```python
#Qubit gates

X = [[0, 1], [1, 0]]
print("X =", X)

Y = [[0, complex(0, -1)], [complex(0, 1), 0]]
print("Y =", Y)

Z = [[1, 0], [0, -1]]
print("Z =", Z)
```

```python
#Hadamard gate

H = [[1/(2**0.5), 1/(2**0.5)], [1/(2**0.5), -1/(2**0.5)]]
print(H)
```

```python
#It is more useful if every element is a complex type so cast all matrices into complex matrices

def complex_matrix(N):
    return [[complex(element) for element in row] for row in N]

c_X = complex_matrix(X)
c_Y = complex_matrix(Y)
c_Z = complex_matrix(Z)
c_H = complex_matrix(H)

print("X =", c_X)
print("Y =", c_Y)
print("Z =", c_Z)
print("H =", c_H)
```

```python
#Make a copy of the Y matrix. Start with a 2×2 matrix of all 0s

Y_copy = [[0, 0], [0, 0]]

for row in range(len(Y)):
    for col in range(len(Y[row])):
        Y_copy[row][col] = Y[row][col]

print(Y_copy)
```

```python
#Create a smiliar cell that makes a transpose of Y, Y^T

Y_transpose = [[0, 0], [0, 0]]

for row in range(len(Y)):
    for col in range(len(Y[row])):
        Y_transpose[col][row] = Y[row][col]

print(Y_transpose)
```

```python
#Now take a Hermitian adjoint and check if Y is 

Y_adjoint = [[0, 0], [0, 0]]

#Calculate hermitian adjoint (transpose + conjugate)
for row in range(len(Y)):
    for col in range(len(Y[row])):
        Y_adjoint[col][row] = Y[row][col].conjugate()

print("Y_adjoint =", Y_adjoint)

#Check if Y is Hermitian
is_hermitian = Y == Y_adjoint
print("Y is Hermitian =", is_hermitian)
```

```python
#Now multiply X and H to get XH

XH = [[0.0, 0.0], [0.0, 0.0]]

for i in range(2):        
    for j in range(2):    
        for k in range(2): 
            XH[i][j] += X[i][k] * H[k][j]

print(XH)
```

```python
#Now multiply XH and H to get HXH

XHX = [[0.0, 0.0], [0.0, 0.0]]

for i in range(2):        
    for j in range(2):    
        for k in range(2): 
            XHX[i][j] += XH[i][k] * X[k][j]

print(XHX)
```
